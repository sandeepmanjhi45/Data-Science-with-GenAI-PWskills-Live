##Question 2: Create a list and tuple in Python, and demonstrate how attempting to change an element works differently for each.


```python
# Creating a list
my_list = [10, 20, 30, 40]

# Displaying the original list
print("Original List:", my_list)

# Changing an element in the list
my_list[1] = 50

# Displaying the modified list
print("Modified List:", my_list)


# Creating a tuple
my_tuple = (10, 20, 30, 40)

# Displaying the original tuple
print("\nOriginal Tuple:", my_tuple)

# Attempting to change an element in the tuple
try:
    my_tuple[1] = 50
except TypeError as e:
    print("Error:", e)
```

#Question 3:What are logical operators in Python?


Logical operators in Python are used to combine two or more conditions and return a Boolean value (True or False).

Python has three main logical operators:
1. AND (and): Returns True if both conditions are true.
2. OR (or): Returns True if at least one condition is true.
3. NOT (not): Reverses the result. If the condition is true, it returns false, and vice versa.

```python
# Example of logical operators in Python

a = 10
b = 5

# AND Operator
print("AND Operator:", a > 5 and b < 10)

# OR Operator
print("OR Operator:", a > 15 or b < 10)

# NOT Operator
print("NOT Operator:", not(a > 5))
```

#Question 4: Write a Python program to convert user input from string to integer, float,and boolean types.


In Python, the input() function always takes user input as a string. We can convert this string into different data types using type conversion functions.

Python provides the following functions:
1. int(): Converts a string into an integer.
2. float(): Converts a string into a floating-point number.
3. bool(): Converts a value into a Boolean value (True or False).

```python
# Taking input from the user
user_input = input("Enter a number: ")

# Converting string into integer
int_value = int(user_input)

# Converting string into float
float_value = float(user_input)

# Converting string into boolean
bool_value = bool(user_input)

# Printing converted values
print("Original String:", user_input)
print("Integer Value:", int_value)
print("Float Value:", float_value)
print("Boolean Value:", bool_value)

# Printing data types
print("\nData Types:")
print(type(int_value))
print(type(float_value))
print(type(bool_value))
```

#Question 5: Describe how dictionaries store data.


 A dictionary in Python is a collection of data stored in
 key-value pairs.

 Each key is unique and is used to access its corresponding value.

 Dictionaries are mutable, which means we can add, modify,
 or delete elements after creation.

### Dictionaries are created using curly brackets {}.

 Example:
 {"name": "Rahul", "age": 21, "city": "Bhopal"}

 Here:
 "name", "age", and "city" are keys.
 "Rahul", 21, and "Bhopal" are values.

```python
# Creating a dictionary
student = {
    "name": "Rahul",
    "age": 21,
    "city": "Bhopal"
}

# Printing the dictionary
print("Student Dictionary:", student)

# Accessing values using keys
print("Name:", student["name"])
print("Age:", student["age"])
print("City:", student["city"])

# Adding a new key-value pair
student["course"] = "Python"

print("Updated Dictionary:", student)
```

#Question 7: What is the difference between for and while loops?


1. For Loop:
- A for loop is used to iterate over a sequence, such as a list, tuple, or string.
- It is generally used when we know the number of iterations or want to process each item in a sequence.
2. While Loop:
- A while loop executes a block of code as long as a given condition is true.
- It is generally used when we do not know the exact number of iterations in advance.

```python
# Example For Loop

print("For Loop:")
for i in range(1, 6):
    print(i)

# Example For While Loop

print("\nWhile Loop:")
i = 1

while i <= 5:
    print(i)
    i += 1
```

#Question 8: Write a Python program to calculate the factorial of a number provided by the user using a while loop.


```python
# Program to calculate factorial using a while loop

# Taking input from the user
num = int(input("Enter a number: "))

# Initializing variables
factorial = 1


# Checking for negative numbers
if num < 0:
    print("Factorial is not defined for negative numbers.")

else:
    # Calculating factorial using For loop
    for i in range(1,num+1):
        factorial = factorial * i
        i += 1

    # Displaying the result
    print("Factorial of", num, "is:", factorial)
```

#Question 9: Why might you use a set instead of a list in Python?

A set in Python is an unordered collection of unique elements. We use a set instead of a list when we want to store only unique values and remove duplicate elements.

Advantages of using a set:
1. Sets automatically remove duplicate values.
2. Sets provide faster membership checking in many cases.
3. Sets support mathematical operations such as union, intersection, and difference.
4. Sets do not maintain a fixed order of elements.

```python
# Creating a list with duplicate elements
my_list = [10, 20, 30, 20, 40, 10]

# Creating a set from the list
my_set = set(my_list)

# Printing the list and set
print("Original List:", my_list)
print("Set after removing duplicates:", my_set)

# Checking an element in the set
print("Is 20 present in set?", 20 in my_set)
```
