```python
# Challenge 2 (Medium): The Todo Manager
#Create a loop that asks the user: "Add task or Remove task? (quit to stop)".

# If "add": Ask for a number and append it to the list.

# If "remove": Ask for a number and remove it from the list.

# If "quit": break the loop.

# Print the list after every change.
```

```python
todo = []

for i in range(1000):
    task = input("Choose add, remove or quit: ")

    if task == 'add':
        num = int(input("Enter number to add: "))
        todo.append(num)
        print(todo)

    elif task == 'remove':
        num = int(input("Enter number to add: "))
        todo.remove(num)
        print(todo)

    elif task == 'quit':
        break

    else:
        print("Invalid Choice, Choose again!")



```

### `super()`

```python
# Super()
# ----- The Parent ----------

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        print('Employee', name, 'created!')

# --------- The Child -------------
# Manager contains name, salary AND one new thing -> team size
class Manager(Employee):
    def __init__(self, name, salary, team_size):
        super().__init__(name, salary)

        self.team_size = team_size

# Super() is used to call the functions (methods) of parent in child class!!
```

```python
my_boss = Manager('Alice', 90000, 5)
```

```python
my_boss.name
```

```python
my_boss.salary
```

```python
my_boss.team_size
```

**Question:** Why do we pass `name` and `salary` to the child class again, when the parent already takes them?

**Answer:** If a child class defines its own `__init__`, it *overrides* the parent's. So the child must accept every value it needs and forward the parent's share with `super().__init__(name, salary)`, then set its own extra attribute (`team_size`). If the child does **not** define `__init__`, Python simply runs the parent's constructor (next cell).

```python
# A child class WITHOUT its own __init__ inherits the parent's constructor
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        print('Employee', name, 'created!')


class Manager(Employee):
    pass   # no __init__ here -> Employee.__init__ runs
```

```python
my_boss = Manager('Alice', 90000)
```

```python
# Scatter plot of plotly
data = {
    "Country": ["India", "USA", "China", "Brazil", "UK", "Nigeria"],
    "Population_M": [1400, 330, 1410, 214, 67, 213],
    "GDP_per_Capita": [2200, 70000, 12500, 7500, 46000, 2000],
    "Continent": ["Asia", "North America", "Asia", "South America", "Europe", "Africa"]
}
df = pd.DataFrame(data)
```

```python
fig = px.scatter(
    df,
    x='GDP_per_Capita', # Horizontal axis
    y='Population_M', # Vertical axis
    color='Continent', # Auto-color by Continent
    size='Population_M', # Make the dot bigger if population is larger
    hover_name = 'Country', # Show the Country name when hovering!
    size_max = 50, # Cap the maximum bubble size so it fits on screen
    title='Global Wealth vs Population'

)

fig.update_layout(
    xaxis_title = 'Wealth (GDP per Capita in USD)',
    yaxis_title = 'Population (in Millions)',
    template = 'plotly_dark'
)

fig.show()
```

```python
# Challenge 2 (Medium): The Car Dealership

# Create a DataFrame:

# Brand: ["Toyota", "Ford", "BMW", "Honda", "Tesla"]

# Horsepower: [150, 250, 300, 140, 400]

# Price: [25000, 35000, 50000, 22000, 60000]

# Type: ["Gas", "Gas", "Gas", "Gas", "Electric"]

# Create an interactive scatter plot: x is Horsepower, y is Price.

# Color the dots by Type.

# Set hover_name to the Brand so the user knows which car they are pointing at.
```

```python
data = {
    "Brand": ["Toyota", "Ford", "BMW", "Honda", "Tesla"],
    "Horsepower": [150, 250, 300, 140, 400],
    "Price": [25000, 35000, 50000, 22000, 60000],
    "Type": ["Gas", "Gas", "Gas", "Gas", "Electric"],
}

df = pd.DataFrame(data)

fig = px.scatter(
    df,
    x='Horsepower',
    y='Price',
    color='Type',
    hover_name='Brand',
)

fig.update_traces(marker=dict(size=15))

fig.show()
```

```python
# Bar chart using plotly

data = {
    "Region": ["North", "North", "South", "South", "East", "East"],
    "Channel": ["Online", "Retail", "Online", "Retail", "Online", "Retail"],
    "Revenue_k": [400, 250, 300, 450, 500, 200]
}
df = pd.DataFrame(data)

fig = px.bar(
    df,
    x='Region',
    y='Revenue_k',
    color='Channel',
    title='Regional Revenue by Sales Channel'
)

#Stacked bar chart

fig.show()
```

```python
# Donut Chart
# Pie Chart (In between you add empty circle it becomes donut chart)

data = {
    "Department": ["Marketing", "Engineering", "HR", "Sales"],
    "Budget_Millions": [12, 45, 5, 20]
}
df = pd.DataFrame(data)

fig = px.pie(
    df,
    names='Department',
    values='Budget_Millions',
    hole=0.5,
    title='Company Budget Allocation'
)

fig.show()
```
